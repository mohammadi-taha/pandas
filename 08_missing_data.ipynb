```python
df = pd.read_csv("movie_scores.csv")
df
```

```python
# Returning NaN values as False:
df.notnull()
```

```python
# Show the data that pre_movie_score is not NaN:
df[df["pre_movie_score"].notnull()]
```

```python
#(1): KEEP DATA: we just keep data as it is. 
```

```python
#(2): DROP DATA: deleting NaN data:
```

```python
# Deleting rows that contains NaN:
df.dropna()
```

```python
# Keeping the rows that at has 2 values at least, delete others: 
df.dropna(thresh=2)
```

```python
# Deleting NaN according to the columns:
df.dropna(axis=1)
```

```python
# If last_name is NaN delete its row:
df.dropna(subset=["last_name"])
```

```python
#(3): REPLACE DATA: fill data with proper data:
```

```python
# Filling with zero:
df["pre_movie_score"] = df["pre_movie_score"].fillna(0)
df
```

```python
# Filling with mean:
df = pd.read_csv("movie_scores.csv")

df["pre_movie_score"] = df["pre_movie_score"].fillna(df["pre_movie_score"].mean())
df
```

```python
# Filling all numeric columns with mean of that column automaticly:
df = pd.read_csv("movie_scores.csv")

df.fillna(df.mean())
```
