```python
import pandas as pd
```

```python
df = pd.read_csv("tips.csv")
df.head()
```

```python
# Selecting one column (type is Series):
df["total_bill"]
```

```python
# Selecting two or more columns (type is DF):
my_col = ["total_bill", "tip"]
df[my_col]
```

```python
# Selecting two or more columns with onother coding(type is DF):
df[["total_bill", "tip"]]
```

```python
# Operation on the columns:
100 * df["tip"] / df["total_bill"] 
```

```python
# Making new column:
df["tip_percentage"] = 100 * df["tip"] / df["total_bill"]
df.head()
```

```python
# Overwrite an existing column:
df["price_per_person"] = df["total_bill"] / df["size"]
df.head()
```

```python
# Rounding values of a column:
import numpy as np
df["price_per_person"] = np.round(df["total_bill"] / df["size"], 2)
df.head()
```

```python
# Removing a column (permanently):
df = df.drop("tip_percentage", axis=1)
df.head()
```
