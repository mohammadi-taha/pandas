```python
import pandas as pd
```

```python
df = pd.read_csv("tips.csv")
df.head()
```

```python
# Add the 'last_four' column to the df, the values should be last four number of CC 'Number':
```

```python
# Checking the data type of the CC Number column:
df.info()
```

```python
# Defining the func:
def last_four(num):
    return str(num)[-4:]
```

```python
# Applying the func to the df and making a new column:
df["last_four"] = df["CC Number"].apply(last_four)
```

```python
df.head()
```

```python
# Using .apply() with multiple columns:
# Find if the tip is Generous according to the total_bill:
import numpy as np

def quality(total_bill, tip):
    if tip / total_bill > 0.20:
        return "Generous"
    else:
        return "Other"
    
df["tip_quality"] = np.vectorize(quality)(df["total_bill"], df["tip"])

df.head()
```
