```python
import pandas as pd
```

```python
one_dict = {"A": ["A0", "A1", "A2"], "B": ["B0", "B1", "B2"]}
two_dict = {"C": ["C0", "C1", "C2"], "D": ["D0", "D1", "D2"]}

one = pd.DataFrame(one_dict)
two = pd.DataFrame(two_dict)
```

```python
one
```

```python
two
```

```python
# Concatinate according to the "columns" (axis=1):
pd.concat([one,two], axis=1)
```

```python
# This is not a good way to do concatinate:
pd.concat([one,two], axis=0)
```

```python
# Add two to the below of the one:
two.columns = one.columns
two
```

```python
pd.concat([one,two], axis=0)
```

```python
# How to fix the index of last concatination:
my_df = pd.concat([one,two], axis=0)
my_df.index = range(len(my_df))
my_df
```
