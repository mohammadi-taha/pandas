```python
messy_names = pd.Series(["Taha   ", "Bo;bo", "   Ali"])
```

```python
messy_names
```

```python
# For the very complicated df, we have to make a func() and using .apply(), otherwise the str methods are good to go:
```

```python
def clean_up(name):
    name = name.replace(";","")
    name = name.strip()
    name = name.capitalize()
    return name
```

```python
messy_names.apply(clean_up)
```
