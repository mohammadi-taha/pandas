```python
import pandas as pd
```

```python
my_index = ["USA", "CANADA", "MEXICO"]
my_data = [1776, 1923, 1854]
```

```python
# Series with default index:
my_ser1 = pd.Series(data=my_data)
my_ser1
```

```python
# Series with labeled index:
my_ser2 = pd.Series(data=my_data, index=my_index)
my_ser2
```

```python
# Making series using dict:
ages = {"Sami": 10, "Hana": 1, "John": 8}
my_ser3 = pd.Series(ages)
my_ser3
```

```python
# Exercise (1): Store employee expenses in a pandas series, don't just manually write in the number, What are Bob's expenses?
import numpy as np
import pandas as pd
import random

emp_index = ["Bob", "Taha", "John"]
emp_expenses = [np.random.randint(1000,1500) for _ in range(3)]

expenses = pd.Series(data=emp_expenses, index=emp_index)
print(expenses)

bob_expense = expenses["Bob"]
print(bob_expense)
```

```python
# Accessing the keys of a Series:
expenses.keys()
```

```python
# Operations:
ages = {"Sami": 10, "Hana": 1, "John": 8}
my_ser4 = pd.Series(ages)
my_ser4 * 2
```

```python
# Operations for multiple series:
sales_q1 = {"Japan": 100, "Iran": 52, "USA": 159}
sales_q2 = {"USA": 178, "Japan": 54, "Mexico": 102}

my_ser5 = pd.Series(sales_q1) 
my_ser6 = pd.Series(sales_q2)

my_ser5 + my_ser6
```

```python
# Not having 'NaN' data:
total_ser = my_ser5.add(my_ser6, fill_value=0)
total_ser
```
