```python
# For DF that aren't in the exact same orders we use 'merge'
```

```python
import pandas as pd
```

```python
registrations = pd.DataFrame({'reg_id':[1,2,3,4],'name':['Andrew','Bobo','Claire','David']})
logins = pd.DataFrame({'log_id':[1,2,3,4],'name':['Xavier','Andrew','Yolanda','Bobo']})
```

```python
registrations
```

```python
logins
```

```python
# inner merge:
pd.merge(registrations,logins,how="inner",on="name")
```

```python
# left merge (order is important):
pd.merge(registrations,logins,how="left",on="name")
```

```python
# right merge (order is important):
pd.merge(registrations,logins,how="right",on="name")
```

```python
# outer merge:
pd.merge(registrations,logins,how="outer",on="name")
```

```python
registrations = registrations.set_index("name")
```

```python
registrations
```

```python
logins
```

```python
# advance selecting the column (on) or index for adding df:
pd.merge(registrations,logins, how="inner", left_index=True, right_on="name")
```

```python
registrations = pd.DataFrame({'reg_id':[1,2,3,4],'name':['Andrew','Bobo','Claire','David']})
logins = pd.DataFrame({'log_id':[1,2,3,4],'name':['Xavier','Andrew','Yolanda','Bobo']})
```

```python
registrations
```

```python
logins
```

```python
registrations = registrations.rename(columns={"name": "reg_name"})
```

```python
registrations
```

```python
logins
```

```python
# advance selecting the column (on) or index for adding df:
result = pd.merge(registrations,logins,how="inner",left_on="reg_name",right_on="name")
```

```python
result
```

```python
# drop the extra column for names:
result = result.drop("reg_name", axis=1)
```

```python
result
```

```python
registrations = pd.DataFrame({'reg_id':[1,2,3,4],'name':['Andrew','Bobo','Claire','David']})
logins = pd.DataFrame({'log_id':[1,2,3,4],'name':['Xavier','Andrew','Yolanda','Bobo']})
```

```python
registrations
```

```python
logins
```

```python
registrations = registrations.rename(columns={"reg_id": "id"})
logins = logins.rename(columns={"log_id": "id"})
```

```python
registrations
```

```python
logins
```

```python
# pandas automatically distinguish two columns with the same name:
pd.merge(registrations,logins,how="inner",on="name")
```

```python
# how to customize the names of the columns with the same names (replace the _x and _y with our own amounts):
pd.merge(registrations,logins,how="inner",on="name", suffixes=("_reg","_log"))
```
