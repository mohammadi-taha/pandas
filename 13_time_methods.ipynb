###### Dealing with European and Americane date style:

```python
# Americane stryle (default): (month-day-year)
# How to convert to European style (day-month-year):
euro_date = "10-12-2000"

# This is not what we want (10 shuld be day):
pd.to_datetime(euro_date)
```

```python
# Fixing the problem:
pd.to_datetime(euro_date,dayfirst=True)
```

###### Dealing with custome style date:

```python
style_date = "12--Dec--2000"
```

```python
# How help pandas to understand that is a datetime and avoiding error:
pd.to_datetime(style_date, format="%d--%b--%Y")
```

###### Pandas is flexible for understanding standard formats of date:

```python
standard_date = "12th of Dec 2000"
```

```python
pd.to_datetime(standard_date)
```

###### 2 ways to convert Series from csv file to datetime dtype:

```python
# First way:
sales = pd.read_csv("RetailSales_BeerWineLiquor.csv")
sales
```

```python
# Checking DATE Series dtype:
sales["DATE"]
```

```python
# changing the DATE Series dtype to datetime:
sales["DATE"] = pd.to_datetime(sales["DATE"])
sales["DATE"]
```

```python
# Second way (faster way):
sales = pd.read_csv("RetailSales_BeerWineLiquor.csv", parse_dates=[0])
sales["DATE"]
```

###### resample:

```python
# It is groupby that works with date:
sales = pd.read_csv("RetailSales_BeerWineLiquor.csv", parse_dates=[0])
sales
```

```python
sales = sales.set_index("DATE")
sales
```

```python
sales.resample(rule="A").mean()
```
